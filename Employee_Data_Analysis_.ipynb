Create employee data

```python
data = {
    "Employee_ID": range(101, 116),

    "Name": [
        "Ravi", "Priya", "Arun", "Sneha", "Kiran",
        "Divya", "Manoj", "Anu", "Rahul", "Meena",
        "John", "Kavya", "Ajay", "Swathi", "Varun"
    ],

    "Department": [
        "IT", "HR", "IT", "Finance", "IT",
        "HR", "Finance", "IT", "Marketing", "Marketing",
        "IT", "HR", "Finance", "IT", "Marketing"
    ],

    "Age": [
        24, 26, 29, 25, 30, 27, 32, 23,
        28, 26, 31, 24, 29, 27, 33
    ],

    "Salary": [
        45000, 35000, 65000, 40000, 75000,
        42000, 55000, 38000, 48000, 52000,
        80000, 36000, 60000, 58000, 70000
    ],

    "Experience": [
        2, 3, 5, 2, 7, 4, 8, 1,
        4, 3, 9, 2, 6, 4, 10
    ]
}

df = pd.DataFrame(data)

print("Employee dataset created successfully!")
```

**Data Cleaningg**


Introduce missing values

```python
df.loc[2, "Age"] = np.nan
df.loc[5, "Salary"] = np.nan
df.loc[8, "Experience"] = np.nan

df
```

Identify missing values

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

Handle missing values

```python
df["Age"] = df["Age"].fillna(df["Age"].median())

df["Salary"] = df["Salary"].fillna(df["Salary"].median())

df["Experience"] = df["Experience"].fillna(
    df["Experience"].median()
)

print(df.isnull().sum())
```

Introduce and remove duplicates

```python
df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

print("Before removing duplicates:", len(df))

df = df.drop_duplicates()

print("After removing duplicates:", len(df))
```

Check the cleaned dataset

```python
print("Final dataset shape:", df.shape)
print("Missing values:")
print(df.isnull().sum())

df.head(15)
```

Basic employee statistics

```python
print("Total Employees:", df["Employee_ID"].nunique())

print("Average Salary:", df["Salary"].mean())

print("Median Salary:", df["Salary"].median())

print("Highest Salary:", df["Salary"].max())

print("Lowest Salary:", df["Salary"].min())

print("Average Age:", df["Age"].mean())

print("Average Experience:", df["Experience"].mean())
```

Find the highest-paid employee

```python
highest_paid = df.loc[df["Salary"].idxmax()]

print("Highest Paid Employee:")
print(highest_paid[["Name", "Department", "Salary"]])
```

Department-wise employee count

```python
department_count = df.groupby("Department")["Employee_ID"].count()

print(department_count)
```

Department-wise average salary

```python
department_salary = df.groupby("Department")["Salary"].mean()

print(department_salary.round(2))
```

**NumPy**

Convert salary column into a NumPy array

```python
salary_array = df["Salary"].to_numpy()

print("Salary Array:")
print(salary_array)

print("Data Type:", type(salary_array))
```

Perform statistical calculations

```python
print("Mean Salary:", np.mean(salary_array))

print("Median Salary:", np.median(salary_array))

print("Standard Deviation:", np.std(salary_array))

print("Minimum Salary:", np.min(salary_array))

print("Maximum Salary:", np.max(salary_array))
```

Calculate salary percentiles

```python
percentiles = np.percentile(salary_array, [25, 50, 75])

print("25th Percentile:", percentiles[0])
print("50th Percentile:", percentiles[1])
print("75th Percentile:", percentiles[2])
```

**Data Visualization using Matplotlib**

Department-wise employee count (Bar chart)

```python
department_count = df["Department"].value_counts()

plt.figure(figsize=(8, 5))
department_count.plot(kind="bar")

plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")
plt.xticks(rotation=0)
plt.show()
```

Average salary by department

```python
avg_salary = df.groupby("Department")["Salary"].mean()

plt.figure(figsize=(8, 5))
avg_salary.plot(kind="bar")

plt.title("Average Salary by Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.xticks(rotation=0)
plt.show()
```

Salary distribution (Histogram)

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Salary"], bins=5, edgecolor="black")

plt.title("Employee Salary Distribution")
plt.xlabel("Salary")
plt.ylabel("Number of Employees")

plt.show()
```

Experience vs Salary (Scatter plot)

```python
plt.figure(figsize=(8, 5))

plt.scatter(df["Experience"], df["Salary"])

plt.title("Experience vs Salary")
plt.xlabel("Years of Experience")
plt.ylabel("Salary")

plt.show()
```

**Correlation Analysis**

Calculate correlation

```python
correlation = df[["Age", "Salary", "Experience"]].corr()

print(correlation)
```

Create a correlation heatmap using Matplotlib

```python
plt.figure(figsize=(7, 5))

plt.imshow(correlation, cmap="coolwarm")

plt.colorbar()

plt.xticks(range(len(correlation.columns)), correlation.columns)
plt.yticks(range(len(correlation.columns)), correlation.columns)

for i in range(len(correlation.columns)):
    for j in range(len(correlation.columns)):
        plt.text(j, i, round(correlation.iloc[i, j], 2),
                 ha="center", va="center")

plt.title("Employee Data Correlation Heatmap")
plt.show()
```
